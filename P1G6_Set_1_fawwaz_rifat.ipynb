Berdasarkan Uji Skewness diatas, diketahui bahwa tidak ada fitur yang memiliki distribusi yang normal, hal ini sesuai dengan kurva diatas dimana tidak ada fitur yang memiliki bentuk lonceng. Semua fitur pada data ini akan dilakukan capping terlebih dahulu dengan menggunakan Z-Sores untuk fitur Gaussian dan Tukey's Rule unutk fitur yang tidak Gaussian. Fitur yang tidak berdistribusi normal adalah fitur yang memiliki nilai diluar range -0,5 - 0,5. Fitur yang memimilki nilai skew diluar range -1 - 1 dianggap sebagai data yang highly skewed dana akan menggunakan nilai nilai fold = 3 yang artinya nilai IQRnya akan dikalikan 3.

```python
gaussian = ['PURCHASES_FREQUENCY', 'PURCHASES_INSTALLMENTS_FREQUENCY']
high_skew = ['BALANCE', 'BALANCE_FREQUENCY', 'PURCHASES','ONEOFF_PURCHASES', 'INSTALLMENTS_PURCHASES', 'CASH_ADVANCE','ONEOFF_PURCHASES_FREQUENCY', 'CASH_ADVANCE_FREQUENCY',
'CASH_ADVANCE_TRX', 'PURCHASES_TRX', 'CREDIT_LIMIT', 'PAYMENTS','MINIMUM_PAYMENTS', 'PRC_FULL_PAYMENT']

# Winsorizer untuk fitur dengan distribusi normal
gaussian_winsoriser = Winsorizer(capping_method='gaussian', tail='both', fold=3, variables= gaussian)

# Winsorizer untuk fitur dengan distribusi high skew
high_skew_winsoriser = Winsorizer(capping_method='iqr', tail='both', fold=3, variables=high_skew)

# Fit dan transform fitur gaussian
gaussian_winsoriser.fit(data_1)
data_1 = gaussian_winsoriser.transform(data_1)

# Fit dan transform fitur high skew
high_skew_winsoriser.fit(data_1)
data_1 = high_skew_winsoriser.transform(data_1)
```

```python
# Uji Distribusi Normalitas Skewness
print('Nilai Normalitas Skewness BALANCE                            : ', data_1['BALANCE'].skew())
print('Nilai Normalitas Skewness BALANCE_FREQUENCY                  : ', data_1['BALANCE_FREQUENCY'].skew())
print('Nilai Normalitas Skewness PURCHASES                          : ', data_1['PURCHASES'].skew())
print('Nilai Normalitas Skewness ONEOFF_PURCHASES                   : ', data_1['ONEOFF_PURCHASES'].skew())
print('Nilai Normalitas Skewness INSTALLMENTS_PURCHASES             : ', data_1['INSTALLMENTS_PURCHASES'].skew())
print('Nilai Normalitas Skewness CASH_ADVANCE                       : ', data_1['CASH_ADVANCE'].skew())
print('Nilai Normalitas Skewness PURCHASES_FREQUENCY                : ', data_1['PURCHASES_FREQUENCY'].skew())
print('Nilai Normalitas Skewness ONEOFF_PURCHASES_FREQUENCY         : ', data_1['ONEOFF_PURCHASES_FREQUENCY'].skew())
print('Nilai Normalitas Skewness PURCHASES_INSTALLMENTS_FREQUENCY   : ', data_1['PURCHASES_INSTALLMENTS_FREQUENCY'].skew())
print('Nilai Normalitas Skewness CASH_ADVANCE_FREQUENCY             : ', data_1['CASH_ADVANCE_FREQUENCY'].skew())
print('Nilai Normalitas Skewness CASH_ADVANCE_TRX                   : ', data_1['CASH_ADVANCE_TRX'].skew())
print('Nilai Normalitas Skewness PURCHASES_TRX                      : ', data_1['PURCHASES_TRX'].skew())
print('Nilai Normalitas Skewness CREDIT_LIMIT                       : ', data_1['CREDIT_LIMIT'].skew())
print('Nilai Normalitas Skewness PAYMENTS                           : ', data_1['PAYMENTS'].skew())
print('Nilai Normalitas Skewness MINIMUM_PAYMENTS                   : ', data_1['MINIMUM_PAYMENTS'].skew())
print('Nilai Normalitas Skewness PRC_FULL_PAYMENT                   : ', data_1['PRC_FULL_PAYMENT'].skew())
print('Nilai Normalitas Skewness TENURE                             : ', data_1['TENURE'].skew())
```

```python
# Mmebuat Plot Cummulative of Explained Variance Ratio
plt.figure(figsize=(15, 5))
plt.subplot(1, 2, 1)
plt.grid()
cumulative_variance = np.cumsum(pca.explained_variance_ratio_ * 100)
plt.plot(range(1, len(cumulative_variance) + 1), cumulative_variance)
plt.hlines(95, 1, len(cumulative_variance), colors='r', linestyles='dashed')
plt.vlines(11, min(cumulative_variance), max(cumulative_variance), colors='g', linestyles='dashed')
plt.xlabel('Number of components')
plt.ylabel('Explained Variance Ratio - Cummulative')

# Mmebuat Plot Eigenvalues
plt.subplot(1, 2, 2)
plt.grid()
plt.plot(range(1, len(pca.explained_variance_) + 1), pca.explained_variance_)
plt.xlabel('Number of components')
plt.ylabel('Eigenvalues')
```

Berdasarkan visualisasi dari Explained Variance Ratio (0,95) didapatkan jumlah komponen setelah PCA berjumlah 11, tetapi dengan margin of error 1, sehingga perlu dilakukan metode lain yang lebih objektif. Selain itu, dari visualisasi Eigenvalues didapatkan jumlah komponen 12. Namun, hasil visualisasi ini masih subjektif, sehingga perlu diuji dengan metode yang lebih objektif.

```python
# Melakukan PCA untuk menentukan jumlah fitur yang akan dimasukkan dalam model

cumsum = np.cumsum(pca.explained_variance_ratio_)
num_features = np.argmax(cumsum >= 0.95) + 1
num_features
```

## **ix. Model Evaluation**

```python
# Mencari jumlah Elbow Method
k_inertia = []
#membuat iterasi dari 2-10 untuk nilai k terbaik
for i in range(2, 11):
    km = KMeans(n_clusters = i, init = 'k-means++', max_iter = 300, n_init = 10, random_state = 3)
    km.fit(data_scaled_pca)
    k_inertia.append(km.inertia_)

plt.plot(range(2, 11), k_inertia, marker ="o")
plt.title('The Elbow Method', fontsize = 20)
plt.xlabel('No. of Clusters')
plt.ylabel('k_inertia')
plt.show()
```

Berdasarkan visualisasi Elbow Method tersebut maka kemungkinan jumlah klaster terbaik dimulai dari 3-10 karena mulai pada range angka tersebut tidak terjadi penurunan k_inertia yang begitu besar, sehingga harus diuji klaster tersebut satu per satu untuk mendapatkan jumlah klaster terbaik.

```python
def plot_silhouette(range_n_clusters, X, random_state):
    for n_clusters in range_n_clusters:
        # Create a subplot with 1 row and 2 columns
        fig, (ax1, ax2) = plt.subplots(1, 2)
        fig.set_size_inches(18, 4)

        # The 1st subplot is the silhouette plot
        # The silhouette coefficient can range from -1, 1 but in this example all lie within [-0.1, 1]
        ax1.set_xlim([0, 1])
        # The (n_clusters+1)*10 is for inserting blank space between silhouette plots of individual clusters, to demarcate them clearly.
        ax1.set_ylim([0, len(X) + (n_clusters + 1) * 10])

        # Initialize the clusterer with n_clusters value and a random generator seed of 10 for reproducibility.
        clusterer = KMeans(n_clusters = n_clusters, init = 'k-means++', max_iter = 300, n_init = 10, random_state = random_state)
        cluster_labels = clusterer.fit_predict(X)

        # The silhouette_score gives the average value for all the samples.
        # This gives a perspective into the density and separation of the formed clusters
        silhouette_avg = silhouette_score(X, cluster_labels)
        print("For n_clusters =", n_clusters,
              "The average silhouette_score is :", silhouette_avg)

        # Compute the silhouette scores for each sample
        sample_silhouette_values = silhouette_samples(X, cluster_labels)

        y_lower = 10
        for i in range(n_clusters):
            # Aggregate the silhouette scores for samples belonging to cluster i, and sort them
            ith_cluster_silhouette_values = sample_silhouette_values[cluster_labels == i]
            ith_cluster_silhouette_values.sort()

            size_cluster_i = ith_cluster_silhouette_values.shape[0]
            y_upper = y_lower + size_cluster_i

            color = cm.nipy_spectral(float(i) / n_clusters)
            ax1.fill_betweenx(np.arange(y_lower, y_upper),
                              0, ith_cluster_silhouette_values,
                              facecolor=color, edgecolor=color, alpha=0.7)

            # Label the silhouette plots with their cluster numbers at the middle
            ax1.text(-0.05, y_lower + 0.5 * size_cluster_i, str(i))

            # Compute the new y_lower for next plot
            y_lower = y_upper + 10  # 10 for the 0 samples

        ax1.set_title("The silhouette plot for the various clusters.")
        ax1.set_xlabel("The silhouette coefficient values")
        ax1.set_ylabel("Cluster label")

        # The vertical line for average silhouette score of all the values
        ax1.axvline(x=silhouette_avg, color="red", linestyle="--")

        ax1.set_yticks([])  # Clear the yaxis labels / ticks
        ax1.set_xticks([0, 0, 0.2, 0.4, 0.6, 0.8, 1])

        # 2nd Plot showing the actual clusters formed
        colors = cm.nipy_spectral(cluster_labels.astype(float) / n_clusters)
        ax2.scatter(X[:, 0], X[:, 1], marker='.', s=30, lw=0, alpha=0.7,
                    c=colors, edgecolor='k')

        # Labeling the clusters
        centers = clusterer.cluster_centers_

        # Draw white circles at cluster centers
        ax2.scatter(centers[:, 0], centers[:, 1], marker='o',
                    c="white", alpha=1, s=200, edgecolor='k')

        for i, c in enumerate(centers):
            ax2.scatter(c[0], c[1], marker='$%d$' % i, alpha=1, s=50, edgecolor='k')

        ax2.set_title("The visualization of the clustered data.")
        ax2.set_xlabel("Feature space for the 1st feature")
        ax2.set_ylabel("Feature space for the 2nd feature")

        plt.suptitle(("Silhouette analysis for KMeans clustering on sample data "
                      "with n_clusters = %d" % n_clusters),
                     fontsize=14, fontweight='bold')
```

```python
# Menampilkan Silhouette Score dan Silhouette Plot

plot_silhouette(range(3, 11), data_scaled_pca, 7)
```

```python
#Silhoutte Plot Analysis

#diatur mau dicoba berapa cluster disini
fig, axes = plt.subplots(2, 6, figsize=(15, 8))

# Loop melalui setiap subplot
for i, ax in enumerate(axes.flatten()):
    # Set jumlah cluster
    n_clusters = i + 2

    # Initialize K-means clustering
    kmeans = KMeans(n_clusters=n_clusters, init="k-means++", n_init = 10, max_iter = 100, random_state = 42)

    # Buat SilhouetteVisualizer
    visualizer = SilhouetteVisualizer(kmeans, colors='yellowbrick', ax=ax)

    # Fit visualizer ke data x
    visualizer.fit(data_scaled_pca)

    # Set title
    ax.set_title(f'K = {n_clusters}')

# Adjust layout
plt.tight_layout()
plt.show()
```

Berdasarkan Silhouette Score dan Silhouette Plot di atas, maka jumlah klaster yang diambil dan akan digunakan untuk melatih model adalah n_clusters = 4. Klasterisasi 4 dipilih karena beberapa alasan, yaitu semua klasternya memiliki Silhouette Score diatas nilai rata-ratanya, Silhoutte Plot/ jumlah data tiap klasternya tidak terlalu timpang, dan memiliki sebaran/scatterplot yang lebih baik dibandingkan klaster lain karena tidak terdapat terlalu banyak overlap pada data. n_clusters = 3 tidak dipilih karena jumlah data tiap klasternya terlalu timpang/tidak seimbang dimana klaster 2 memiliki jumlah data yang jauh lebih banyak dibandingkan klaster lainnya. Oleh karena itu, berdasarkan beberapa alasan yang telah dijelaskan, maka n_clusters = 4 dipilih untuk dapat dapat digunakan dalam pembuatan model.
